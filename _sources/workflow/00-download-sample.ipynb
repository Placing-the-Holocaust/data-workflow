```python
from huggingface_hub import hf_hub_download, list_repo_files
import os
import random
import PyPDF2
import io
from PIL import Image
```

```python
repo_id = "placingholocaust/ushmm-pdfs"
output_dir = "yolo"
os.makedirs(output_dir, exist_ok=True)
image_dir = "RG-50.999.0561_trs_en"
os.makedirs(image_dir, exist_ok=True)

# Download the PDF file
hf_hub_download(repo_id="placingholocaust/ushmm-pdfs", filename=f"pdfs/{image_dir}.pdf", local_dir=output_dir, repo_type="dataset")

# Extract the filename from the path
filename = os.path.basename(f"pdfs/{image_dir}.pdf")

# Define the local path where the file is saved
local_path = os.path.join(output_dir, "pdfs", filename)

print(f"Downloaded: {filename}")
print(f"Saved to: {local_path}")

# Open the PDF and convert all pages to images
with open(local_path, 'rb') as file:
    reader = PyPDF2.PdfReader(file)
    num_pages = len(reader.pages)
    print(f"Number of pages: {num_pages}")
    
    # Convert all pages to images
    from pdf2image import convert_from_path
    images = convert_from_path(local_path)
    
    for page_num, image in enumerate(images, start=1):
        # Generate a unique filename for each image
        image_filename = f"{os.path.splitext(filename)[0]}_page{page_num:04}.png"
        image_path = os.path.join(image_dir, image_filename)
        
        # Save the image
        image.save(image_path)
        
        print(f"Converted page {page_num} to image: {image_filename}")
        print(f"Saved to: {image_path}")

print(f"All pages from {filename} have been converted to images and saved in {image_dir}")

```
