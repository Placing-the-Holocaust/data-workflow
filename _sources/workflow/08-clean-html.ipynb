```python
import os
from collections import Counter
from tqdm import tqdm
import re

with open("../assets/remove_phrases.txt", "r") as f:
    phrases_to_remove = f.read().splitlines()

# Create the output directory if it doesn't exist
os.makedirs("../data/05_final_removals", exist_ok=True)

for file in tqdm(files):
    with open(file, "r") as f:
        soup = BeautifulSoup(f, "html.parser")
    
    sentence_tags = soup.find_all("sentence")
    
    # Process each sentence
    for sentence in sentence_tags:
        text = sentence.get_text().strip()
        
        for phrase in phrases_to_remove:
            text = text.replace(phrase, '')
        
        text = " ".join(text.split())
        
        # Update the sentence text in the soup
        if len(text) > 1:
            sentence.string = text
        else:
            sentence.decompose()  # Remove empty sentences
    
    # Get the original filename and create new path
    filename = os.path.basename(file)
    output_path = os.path.join("../data/05_final_removals", filename)
    
    # Save the modified HTML
    with open(output_path, "w") as f:
        f.write(str(soup))
```
