```python
from bs4 import BeautifulSoup
import glob
```

```python
files = glob.glob("../data/05_final_removals/*.html")
files.sort()
len(files)
http_count = 0
for file in files:
    with open(file, "r") as f:
        soup = BeautifulSoup(f, "html.parser")
    found = False
    sentences = soup.find_all("sentence")
    for sentence in sentences:
        if "http://collections.ushmm.org" in sentence.text:
            found = True
            http_count += 1
            
    if found:  # Only print if the text was actually found
        print(f"Found in {file}:")

print(f"\nTotal files with http reference: {http_count}")
```
